```python
#importing the packages
import igraph as ig
import matplotlib.pyplot as plt
```

```python
# 1. Initialize graph and add the 10 vertices
g = ig.Graph()
nodes = ["Andre", "Beverly", "Carol", "Diane", "Ed", "Fernando", "Garth", "Heather", "Ike", "Jane"]
g.add_vertices(nodes)

# 2. Add the connections (Edges)
edges = [
    ("Andre", "Beverly"), ("Andre", "Carol"), ("Andre", "Diane"), ("Andre", "Fernando"),
    ("Beverly", "Diane"), ("Beverly", "Ed"), ("Beverly", "Garth"),
    ("Carol", "Diane"), ("Carol", "Fernando"),
    ("Ed", "Diane"), ("Ed", "Garth"),
    ("Diane", "Fernando"), ("Diane", "Garth"),
    ("Fernando", "Garth"), ("Fernando", "Heather"),
    ("Garth", "Heather"),
    ("Heather", "Ike"),
    ("Ike", "Jane")
]
g.add_edges(edges)

# 3. Define manual (x, y) layout coordinates to form the Kite shape
# (0,0 is top-left in igraph's plotting system)
layout_coordinates = [
    (1, 2),  # Andre
    (1, 4),  # Beverly
    (2, 1),  # Carol
    (2, 3),  # Diane
    (2, 5),  # Ed
    (3, 2),  # Fernando
    (3, 4),  # Garth
    (4, 3),  # Heather
    (5, 3),  # Ike
    (6, 3)   # Jane
]
graph_layout = ig.Layout(layout_coordinates)

# 4. Styling properties to match your image
visual_style = {
    "layout": graph_layout,
    "vertex_shape": "rectangle",       # Matches the square icons
    "vertex_size": 24,
    "vertex_color": "#d3d3d3",         # Light gray node color
    "vertex_label": g.vs["name"],      # Displays names
    "vertex_label_size": 14,
    "vertex_label_color": "black",
    "vertex_label_dist": 1.5,          # Pushes text slightly away from the node
    "edge_color": "#800080",           # Purple/dark magenta edge color
    "edge_width": 3,                   # Thick edge lines
    "bbox": (600, 400),
    "margin": 50
}

# 5. Plot the graph
ig.plot(g, **visual_style)
```
